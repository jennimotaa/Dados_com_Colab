```python
import math

def somar (a,b):
  return a+b

def subtrair (a,b):
  return a-b

def multiplicar (a,b):
  return a*b

def dividir (a,b):
  if b==0:
    return "Erro: divisão por zero"
  return a/b

def potencia (a,b):
  return a**b

def raiz_quadrada (a):
  if a<0:
    return "Erro: não existe raiz quadrada de número negativo!"
  return math.sqrt(a)

continuar = "S"

while continuar.upper()== "S":
  print("--Calculadora--")

  n1 = float (input("Digite o primeiro número: "))
  n2 = float (input("Digite o segundo número: "))

  print("Escolha a operação:")

  print("+ Soma")
  print("- Subtração")
  print("* Multiplicação")
  print("/ Divisão")
  print("** Potência")
  print("// Raiz quadrada (apenas do primeiro número)")

  operacao = input("Qual a operação: ")

  if operacao == "+":
    print("Resultado =", somar(n1,n2))

  elif operacao == "-":
    print("Resultado =", subtrair(n1,n2))

  elif operacao == "*":
    print("Resultado =", multiplicar(n1,n2))

  elif operacao == "/":
    print("Resultado =", dividir(n1,n2))

  elif operacao == "**":
    print("Resultado =", potencia(n1,n2))

  elif operacao == "//":
    print("Resultado =", raiz_quadrada(n1))
  else:
    print("Operação inválida!")

  continuar = input("Deseja continuar? (S/N): ")

print("Calculadora encerrada")
```
